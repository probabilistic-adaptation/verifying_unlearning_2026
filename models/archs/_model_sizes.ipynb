# Seeing model sizes

```python

from models.archs.utils import init_model
```

```python
resnet_model = init_model(model_class = "ResNet", num_classes = 10, checkpoint_path = None).to("cpu")
allcnn_model = init_model(model_class = "AllCNN", num_classes = 10, checkpoint_path = None).to("cpu")
```

```python
print(f"Resnet: {sum(p.numel() for p in resnet_model.parameters()):,}")
print(f"AllCNN: {sum(p.numel() for p in allcnn_model.parameters()):,}")
```
