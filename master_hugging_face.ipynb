# Download (pull specific seeds only)

```python
from huggingface_hub import snapshot_download

# edit this to whichever seed(s) you just pushed
seeds = [4, 5, 6, 11]

patterns = []
for seed in seeds:

    print(f"Downloading results and model checkpoints for seed {seed}...")
    
    patterns += [
        f"results/seed_{seed}/*",
        f"models/model_checkpoints/seed_{seed}/*",
    ]

snapshot_download(
    repo_id="jjmoncus1/verifying_unlearning_2026",
    repo_type="dataset",
    allow_patterns=patterns,   # restricts download to just these subdirs
    local_dir=".",             # mirrors the repo layout into the project root
)
```
